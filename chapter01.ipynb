### 변수 이름 규칙
- 숫자로 시작 불가 : 1_name(x)
- 공백 불가 : user name (x)
- 하이픈(-) : user-age (x)
- 예약어 사용불가 : if, def (x)

### 컨벤션
- 스네이크 표기법 : user_name, value_list

### 다른 표기법
- 파스칼 케이스 : UserName, VallueList
- 카멜 케이스 (단봉 낙타법) : userName

````python
# 정수형 자료형
age = 25
# 실수 자료형
pi = 3.14
# 불리언 자료형 (Boolean)
is_alive = True
# 문자 자료형
user_name = "SinYuno"
# 배열 Array
grade_type = [1, 2, 3, 4]
````
