## 종합 실습

```python
# 회원 데이터
user_database = [
    {"name":"손흥민", "level":"VIP", "active":True, "point":15000}, 
    {"name":"김민재", "level":"Gold", "active":False, "point":9000}, 
    {"name":"조규성", "level":"VIP", "active":True, "point":12000}, 
    {"name":"이강인", "level":"Gold", "active":True, "point":8000}, 
    {"name":"홍명보", "level":"Bronze", "active":False, "point":1000}
]

# VIP 포인트 누적 변수
total_vip_points = 0
# 현재 활성 회원 수
active_user_count = 0

print("===== 마케팅 대상 활성 VIP 및 Gold 회원 결과 =====")

for user in user_database:
    # 활성화 된 고객 필터링 active
    if user["active"]:
        # 활성 회원 수 누적
        active_user_count += 1

        if user["level"] == "VIP":
            total_vip_points += user["point"]

            print(f"[VIP 등급] {user['name']}회원님에게 VIP 쿠폰이 발행되었습니다.")
        elif user["level"] == "Gold":
            print(f"[Gold 등급] {user['name']}회원님에게 Gold 쿠폰이 발행되었습니다.")

print(f"전체 활성 회원 수 : {active_user_count}명")
print(f"활성 VIP 포인트 누적 : {total_vip_points}점")
```

### 숫자 맞추기 게임

```python
# 난수 발생을 위한 모듈 임포트
import random 

# 1 ~ 100 사이의 난수 발생
number = random.randint(1,10)

# 시도 횟수
try_count = 0

# 타이틀 
print("=" * 46)
print("      컴퓨터가 1부터 10 사이의 숫자를 생성했습니다.")
print("=" * 46)

# 메인 루프

while True:
    user_input = input("컴퓨터의 숫자를 맞춰보세요. (1~10): ")

    if user_input == "q":
        print(f"게임을 종료합니다. 정답은 {number}입니다.")
        break
    
    user_number = int(user_input)

    if user_number > 10 or user_number < 1:
        print("1 ~ 10 사이의 숫자만 입력해주세요.")
        continue
    if not user_input.isdecimal():
        print("숫자만 입력해주세요.")
        continue

    # 시도 횟수 증가
    try_count += 1

    # 입력값과 정답을 비교
    if user_number < number:
        print("Up! 더 큰 수 입니다.")
    elif user_number > number:
        print("Down! 더 작은 수 입니다.")
    else:
        print(f"정답입니다. {try_count}회 만에 성공 했습니다.")
        break # 정답일 경우 무한루프 탈출
```
