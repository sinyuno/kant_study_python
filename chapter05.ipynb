### for 반복문

```python
for 변수 in 반복할_데이터:
    반복수행할 코드 #1
    반복수행할 코드 #2
```

````python
# 리스트
fruits = ["사과", "바나나", "체리"]

for fruit in fruits:
    print(fruit)
````

````python
for ch in "스파르타":
    print(ch)
````

````python
menu = {"아아":4000, "홍차":4500}

menu.keys()
for key in menu:
    print(key)

for name, price in menu.items():
    print(f"{name} : {price}")
````

````python
# range() 기본 사용법

for i in range(5):
    print(f"{i + 1} 번째 로직 수행")
````

````python
# 범위 지정

for i in range(1, 6, 2): # 1부터 6 전까지 2씩 띄어서 반복
    print(i)

print(range(5))
print(list(range(5)))  # 리스트로 변환 후 출력
````

### while 문

```python
while 조건식:
    반복할 로직 #1
    반복할 로직 #2
    반복할 로직 #3
```
탈출 루트를 만들지 않으면 무한루프에 빠지기 때문에 강제 종료를 해야 할 수도 있음.

````python
num = 1

while num <= 3: # num이 3보다 작거나 같을 때 까지만 반복
    print(f"현재 숫자는 {num}")
    num += 1 # num이 반복 될 때마다 1씩 더하기
````

````python
count = 3

while count > 0:
    print(count)
    count -= 1 # count = count - 1
````

````python
for num in range(1, 6):
    if num == 3:
        print("즉시 종료")
        break
    print(f"현재 num={num}")
````

````python
while True:
    user_input = input()
    if user_input == "q":
        print("종료")
        break
````

````python
for num in range(1, 6):
    if num % 2 == 0: # 짝수
        continue
    print(f"홀수 {num}")
````

````python
# 구구단

for x in range (2, 10):
    if x == 4:
        break
    print(f"--- {x}단 ---")
    for y in range(1, 10):
        if y == 6:
            break
        print(f"{x} x {y} = {x * y}")
````
