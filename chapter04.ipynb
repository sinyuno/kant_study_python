## 딕셔너리 (Dictionary)

- {"키":"값"}
- "키" 중복 불가
- "값" 중복 가능
- 추가하는 순서를 유지

```python
#print({"a" : 1, "b" : 2})
print({"a" : 1, "b" : 1})
```

### 추가 수정

```python
user_info = {
    "name" : "SinYuno",
    "age" : 24
}

print(user_info)
print(user_info["age"]) # 키로 값을 가져오기
user_info["age"] = 30
print(user_info["age"])
user_info["role"] = "Tutor"
print(user_info)

# user_info[0]
```

### 삭제

```python
user_info = {
    "name" : "SinYuno",
    "age" : 24,
    "role" : "admin"
}
print(user_info)
# del 메서드 사용
del user_info["role"] # 키 삭제

print(user_info)

# pop 사용 : 꺼내면서 삭제
age = user_info.pop("age") # 키 삭제 후 값 반환
print(age, user_info)

#존재 하지 않는 값을 pop으로 꺼내려고 하면 KeyError 발생
```

### 딕셔너리 주요 메서드

- Keys(), values(), items()


```python
menu = {"coffee" : 4000, "tea" : 4500}

# Key 목록 조회
print(menu.keys()) # dict_keys(['coffee', 'tea'])

# Value 목록 조회
print(menu.values()) # dict_values([4000, 4500])

# items 조회 (키, 값) 튜플 형태로 리턴
print(menu.items()) # dict_items([('coffee', 4000), ('tea', 4500)])

# 딕셔너리는 인덱서 방식으로 값을 참조 불가 > list로 변환
print(list(menu.keys())[0])

print(sum(menu.values())) # 값들의 합계
```

```python
print("coffee" in menu) # True
print(4000 in menu) # False

print(4000 in menu.values()) # True
```

```python
# menu["cake"] -> 없는 값을 가져오기 때문에 KeyError 발생
price = menu.get("cake", 5000)

print(price)

# default 값이 없는 경우 None 반환
print(menu.get("cake"))

print(menu.get("coffee", 0))

print(menu)
```

### 집합 (Set)

수학 집합 : 중복 없고, 순서 없는 자료형

- 중괄호 사용
- set() 함수를 생성

```python
# set은 중복이 없기 때문에 1, 2, 3만 출력, 순서가 없기 때문에 인덱스 활용 출력 불가

grade = set([1, 2, 2, 3, 3, 3])

print(grade)
```

```python
print(type({})) # {} 빈 딕셔너리
print(type(set())) # set() 빈 집합
print({1, 2, 2}) # 중괄호 활용 집합 생성
```

```python
s = {1, 2}

s.add(2) # 집합은 중복이 불가하므로 추가 안됨.
s.add(5)

print(s)

s.remove(1) # 값이 없는것을 삭제 하려 할 때 오류 발생
print(s)

s.discard(2) # 값이 없는것을 삭제 하려 할 때 오류 발생하지 않음.
print(s)
```

```python
set_a = {1, 2, 3}
set_b = {3, 4, 5}

print(set_a & set_b) # 교집합
print(set_a | set_b) # 합집합 
print(set_a - set_b) # 차집합
print(3 in set_a) # 포함 여부
```

### 중복제거

```python
# 원시 데이터
raw_emails = ["a@sparta.com", "b@sparta.com", "a@sparta.com"]
# 집합으로 중복 제거
unique_set = set(raw_emails)

print(list(unique_set))

print("# 중복 제거 및 순서 유지")
sort_emails = list(dict.fromkeys(raw_emails))
print(sort_emails)

# sorted
# 역순은 sorte의 reverse 값 True로 변경
print(sorted(set(["c", "d", "a", "f", "a"])))
```
