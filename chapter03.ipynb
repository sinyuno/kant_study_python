````python
# 리스트 : 가변 데이터 저장
shopping_list = ["apple", "banana", "cherry"]

print(shopping_list)
print(shopping_list[0])

shopping_list[0] = "kiwi"
print(shopping_list[0])
print(shopping_list)
````

#### 리스트 메서드

````python
# apend() : 리스트 끝에 데이터 추가
fruits = ["apple"]
fruits.append("banana")
print(fruits)

# insert() : 리스트 특정 위치에 데이터 추가
fruits.insert(0, "kiwi")
print(fruits)

# remove() : 리스트 특정 데이터 삭제
# fruits.remove("apple")
# print(fruits)

# fruits.remove("grape")
````

### 슬라이싱


````python
numbers = [0, 1, 2, 3, 4, 5]

# 1번 부터 4번 미만까지
print(numbers[1:4])
# 2칸 간격으로 추출
print(numbers[::2])
# 역순으로 추출
print(numbers[::-1])
# 앞에서 3개만 추출
print(numbers[:3])
# 3번부터 끝까지
print(numbers[3:])
````

````python
print((10, 20, 30, 40)[1:3])
````

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
]
print(matrix[1])
print(matrix[1][2])
print(len(matrix), len(matrix[0]))
````
